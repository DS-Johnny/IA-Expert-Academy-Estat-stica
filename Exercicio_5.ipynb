O objetivo desta tarefa é aplicar a distribuição de frequência utilizando o atributo age da base de dados do censo

Carregue a base de dados census.csv
Faça testes utilizando o parâmetro bins para visualizar a distribuição dos dados

```python
import pandas as pd
import numpy as np
import seaborn as sns

dataset = pd.read_csv('Bases de dados/census.csv')
dataset.info()
```

```python
frequencia, classes = np.histogram(dataset['age'], 'sturges')
frequencia, classes, len(classes)
```

```python
sns.histplot(dataset['age'], kde=True)
```

```python
sns.histplot(dataset['age'], kde=True, bins=classes)
```

```python
sns.histplot(dataset['age'], kde=True, bins='sturges')
```

```python
# Recurso do pandas para fazer a divisão de classes pd.cut()
dataset['age'] = pd.cut(dataset['age'], bins=[0, 17, 25, 40, 60, 90], labels=['Faixa1', 'Faixa2', 'Faixa3', 'Faixa4', 'Faixa5'])
dataset.head()
```
