## Implementação Variancia

```python
import numpy as np
from math import sqrt
from scipy import stats
import statistics
```

```python
dados_parcial = np.array([150, 151, 152, 152, 153, 154, 155, 155, 155])
```

```python
media = dados_parcial.mean()
media
```

```python
soma = 0
for i in dados_parcial:
    if (i-media) == 0:
        soma += pow(0,2)

    elif i > media:
        soma += pow(i - media, 2)

    else:
        soma += pow(media - i,2)


variancia = soma/len(dados_parcial)
variancia
```

```python
# Gabarito
desvio = dados_parcial - media
variancia2 = sum(pow(desvio, 2)) / len(dados_parcial)
variancia2
```

```python
# Bibliotecas
print(np.var(dados_parcial))
print(statistics.variance(dados_parcial))
```

```python
from scipy import ndimage
ndimage.variance(dados_parcial)
```

### Desvio Padrão com Dados Agrupados

```python
import pandas as pd
dados = {'inferior' : [150, 154, 158, 162, 166, 170],
         'superior' : [154, 158, 162, 166, 170, 174],
         'fi' : [5, 9, 11, 7, 5, 3]}
dataset = pd.DataFrame(dados)
dataset['xi'] = (dataset['superior'] + dataset['inferior']) / 2 
dataset['fi.xi'] = dataset['fi'] * dataset['xi']
dataset['Fi'] = 0
frequencia_acumulada = []
somatorio = 0
for _,linha in dataset.iterrows():
    #print(linha['inferior'])
    somatorio += linha['fi']
    frequencia_acumulada.append(somatorio)

dataset['Fi'] = frequencia_acumulada
dataset
```

```python
dataset['xi_2'] = dataset['xi'] * dataset['xi']
dataset
```

```python
dataset['fi_xi_2'] = dataset['fi'] * dataset['xi_2']
dataset
```

```python
dataset.columns
```

```python
colunas_orgenadas = ['inferior', 'superior', 'fi', 'xi', 'fi.xi', 'xi_2', 'fi_xi_2', 'Fi']
dataset = dataset[colunas_orgenadas]
dataset
```

```python
dp = sqrt(dataset['fi_xi_2'].sum() / dataset['fi'].sum() - pow(dataset['fi.xi'].sum() / dataset['fi'].sum(), 2))
dp
```
