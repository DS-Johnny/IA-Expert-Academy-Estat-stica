## Implementação Variancia

```python
import numpy as np
from math import sqrt
import statistics
```

```python
dados_parcial = np.array([150, 151, 152, 152, 153, 154, 155, 155, 155])
```

```python
media = dados_parcial.mean()
media
```

```python
soma = 0
for i in dados_parcial:
    if (i-media) == 0:
        soma += pow(0,2)

    elif i > media:
        soma += pow(i - media, 2)

    else:
        soma += pow(media - i,2)


variancia = soma/len(dados_parcial)
variancia
```

```python
# Gabarito
desvio = dados_parcial - media
variancia2 = sum(pow(desvio, 2)) / len(dados_parcial)
variancia2
```

```python
# Bibliotecas
print(np.var(dados_parcial))
print(statistics.variance(dados_parcial))
```
